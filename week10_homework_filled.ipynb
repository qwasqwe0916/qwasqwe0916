## Setup — run this cell without changing it

These packages are already installed in the course `qiskit` environment.

**How to read the results:**

- One **shot** is one repetition of circuit preparation and measurement.
- A result string lists the qubits in the order **q1 q0**: `00` means both values are 0, and `11` means both are 1.
- The four possible result strings are `00`, `01`, `10`, and `11`.
- Counts are obtained from a finite number of shots, so they need not be exactly equal. A fixed seed makes the simulation reproducible.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()
print('Ready to run the Bell-state circuit.')
```

## Q2: Complete the Bell-state circuit and run it. [5 points]

Prepare the Bell state from the lecture:

$$
|\Phi^+\rangle=\frac{|00\rangle+|11\rangle}{\sqrt{2}}.
$$

The circuit starts in $|00\rangle$. Fill in the blanks using these steps:

1. Apply **H to q1**. **(1 point)**
2. Apply **CNOT with control q1 and target q0**. **(1 point)**
3. Measure q0 and q1 into classical bits 0 and 1. **(1 point)**
4. Run the simulation using the variable `SHOTS`, which is set to **100**, and print the counts. **(1 point)**
5. Pass the dictionary `counts_100` to the histogram function and display it. **(1 point)**

### Syntax hints

| Operation | Syntax |
|:---|:---|
| Hadamard gate | `circuit.h(qubit)` |
| CNOT gate | `circuit.cx(control, target)` |
| Measurement | `circuit.measure(qubits, classical_bits)` |
| Simulator repetitions | `simulator.run(compiled, shots=SHOTS, ...)` |
| Histogram | `plot_histogram(count_dictionary, ...)` |

**Required outputs:** the circuit diagram, measurement counts totaling 100, and a histogram.

```python
SHOTS = 100

circuit = QuantumCircuit(2, 2)

# (a) Apply H to q1.
circuit.h(1)

# (b) Apply CNOT: control q1, target q0.
circuit.cx(1, 0)

# (c) Fill in the measurement method name.
circuit.measure([0, 1], [0, 1])
print(circuit.draw())

compiled = transpile(circuit, simulator, seed_transpiler=2026)

# (d) Use the variable that stores the number of shots.
counts = simulator.run(compiled, shots=SHOTS, seed_simulator=2026).result().get_counts()
counts_100 = {s: counts.get(s, 0) for s in ['00', '01', '10', '11']}
print('Number of shots:', SHOTS)
print('Measurement counts:', counts_100)

 
fig, ax = plt.subplots(figsize=(6, 3.5))
plot_histogram(counts_100, ax=ax, title=f'Bell state: {SHOTS} shots')
plt.tight_layout()
plt.show()
```

## Q3: Increase the number of shots and print the probability. [5 points]

Use the same compiled circuit from Q2. Fill in the five blanks below:

1. Set `SHOTS` to **1,000**. **(1 point)**
2. Run the circuit stored in **`compiled`**. **(1 point)**
3. Calculate the measured probability of `00` by dividing its count by the **number of shots**. **(1 point)**
4. Print the probability stored in **`p00`**. **(1 point)**
5. Display the histogram using **`counts_more`**. **(1 point)**

Use

$$
P(00)\approx\frac{\text{number of outcomes labeled 00}}{\text{total number of shots}}.
$$

For this ideal Bell state, the theoretical probability of `00` is 0.5. Your finite-shot estimate should be close to this value, but it need not be exactly 0.5.

**Required outputs:** the number of shots, measurement counts totaling 1,000, the estimated probability and percentage of `00`, and a histogram.

```python
# (a) Set the number of shots to one thousand.
SHOTS = 1000

# (b) Use the compiled circuit from Q2.
counts = simulator.run(compiled, shots=SHOTS, seed_simulator=2026).result().get_counts()
counts_more = {s: counts.get(s, 0) for s in ['00', '01', '10', '11']}
print('Number of shots:', SHOTS)
print('Measurement counts:', counts_more)

# (c) Divide the count of '00' by the number of shots.
p00 = counts_more.get('00', 0) / SHOTS

# (d) Print the variable containing the probability.
print('Estimated P(00) =', p00)
print(f"Percentage of '00': {100 * p00:.1f}%")
 
fig, ax = plt.subplots(figsize=(6, 3.5))
plot_histogram(counts_more, ax=ax, title=f'Bell state: {SHOTS} shots')
plt.tight_layout()
plt.show()
```
